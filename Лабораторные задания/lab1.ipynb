Для набора дискретных данных своего варианта выберите два способа отображения из plot, scatter, stem, step, bar. Объясните, какую особенность данных подчёркивает каждый способ.

```python
import numpy as np
import matplotlib.pyplot as plt

x = list(range(0, 8+1))
y= [0, 0, 1, 1, 2, 2, 3, 3, 4]

# plot строит точки по парам элементов одинакового индекса и соединяет соседние точки линиями
plt.plot(x, y, marker='.') # добавить точки 
plt.show()

# step - cтупенчатое отображение последовательности данных
# step полезен, когда значение интерпретируется как уровень между соседними отсчётами.
plt.step(x, y, where='post')
plt.show()

# если сравнивать эти 2 графика, то наглядно видно, что plot просто соединяет ближайшие точки прямыми линиями, линеаризует между точками функции
# а step показывает ступенчатые изменения значений функции
```

## С7. Три семейства функций и масштаб

Для $P(x)$, $R(x)$ и $T(x)$ своего варианта:
1. вычислите массивы на заданной сетке;
2. корректно обработайте точку $x=d$ рациональной функции;
3. постройте три графика;
4. укажите, является ли особая точка $R(x)$ полюсом или устранимым разрывом;
5. отдельно постройте функцию $G(x)$ своего варианта в обычном и логарифмическом масштабе и сравните представления.

```python
import numpy as np

# 1 Вычисление массивов
x = np.arange(-4, 4+0.25, 0.25)
x_G = np.arange(0, 8+0.4, 0.4)

y_P = 2*x**2 - 3*x - 2
y_T = 1*np.sin(1.5*x+2*np.pi/4) + 2*np.cos(3*x)
y_G = (x_G - 4)**7

print(y_G)
```

```python
# 2 обработка рацинональной функции

y_R = np.full(x.shape, np.nan, dtype=float)
mask = np.abs(x + 1) > 1e-12
y_R[mask] = (-x[mask] + 3) / (x[mask] + 1)
print(y_R)
```

```python
# графики
import matplotlib.pyplot as plt

plt.plot(y_P, label='P(x)')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.show()

plt.plot(y_R, label='R(x)')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.axvline(12, linestyle=':')
plt.show()

plt.plot(y_T, label='T(x)')
plt.grid()
plt.xlabel('x')
plt.ylabel('y')
plt.show()

# 4 так как числитель в точке x = -1 не равен нулю, общего множителя у числителя и знаменателя нет, значит разрыва неустранимый.
```

```python
# 5 функция G(x)

plt.figure(figsize=(7, 6))
plt.plot(x_G, y_G)
plt.scatter(x_G, y_G)
plt.xlabel('x')
plt.ylabel('G(x)')
plt.title('Обычный масштаб')
plt.grid(True)
plt.show()

plt.figure(figsize=(7, 6))
plt.plot(x_G, y_G)
plt.yscale('symlog')
plt.xlabel('x')
plt.ylabel('G(x)')
plt.title('Логарифмический масштаб по y')
plt.scatter(x_G, y_G)
plt.grid(True)
plt.show()
```
